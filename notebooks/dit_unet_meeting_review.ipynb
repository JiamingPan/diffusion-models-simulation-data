### 2. Latest-run source inventory — check this first
Same plans, seed and sample label as `dit_depths_generalization_pca95.ipynb`.
No latest-by-filename guessing. A failed hash check stops execution.
If the plans are unavailable locally, the notebook prints pending status.
On Great Lakes run inside an existing CPU allocation: streaming all references
is substantial disk/CPU work, although no model is run.

```python
from pathlib import Path
import sys, json, hashlib, os
PROJECT_DIR = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p/'simdiff_eval').is_dir()), Path.cwd())
sys.path.insert(0, str(PROJECT_DIR))
BASE = Path('/scratch/huterer_root/huterer0/jiamingp')
PLAN_PATHS = [BASE/'dit_l16_adalnzero_highn_screen_v1/plan.json', BASE/'dit_zero_remaining_300k_v1/plan.json']
SEED, SAMPLE_LABEL = 123, 'dpm50_n512'
EXPECTED_SIZES = [2**k for k in range(6,16)]
def sha256(path):
    h = hashlib.sha256()
    with open(path,'rb') as f:
        for b in iter(lambda:f.read(1024*1024),b''): h.update(b)
    return h.hexdigest()
runs, inventory = [], []
for plan in PLAN_PATHS:
    if not plan.is_file():
        print('UNAVAILABLE PLAN:', plan); continue
    digest = sha256(plan)
    for row in json.loads(plan.read_text())['runs']:
        if int(row['num_layers']) not in (8,12,16) or int(row['patch_size']) != 8: continue
        sample = Path(row['sample_path'].format(seed=SEED,sample_label=SAMPLE_LABEL))
        config = Path(row['config'])
        receipt = sample.with_suffix('.complete.json')
        item = dict(depth=int(row['num_layers']), N=int(row['dataset_size']), run_name=row['run_name'],
                    config=str(config), sample=str(sample), plan=str(plan), plan_sha256=digest)
        if not all(p.is_file() for p in [sample,config,receipt]):
            inventory.append(dict(**item,status='Missing sample/config/receipt')); continue
        saved = json.loads(receipt.read_text())
        assert saved['status']=='complete' and saved['plan_sha256']==digest, receipt
        assert saved['samples_sha256']==sha256(sample), sample
        if row.get('config_sha256'): assert row['config_sha256']==sha256(config), config
        item.update(config_sha256=sha256(config),samples_sha256=saved['samples_sha256'])
        runs.append(item); inventory.append(dict(**item,status='Verified'))
assert len({(r['depth'],r['N']) for r in runs}) == len(runs), 'Duplicate runs: resolve explicitly'
runs.sort(key=lambda r:(r['depth'],r['N']))
if inventory:
    with pd.option_context('display.max_colwidth',None,'display.max_rows',100): display(pd.DataFrame(inventory))
for d in (8,12,16): print('L%d missing sizes:'%d, sorted(set(EXPECTED_SIZES)-{r['N'] for r in runs if r['depth']==d}))
```

### 3. Compute full-reference PDFs and power spectra
All 512 saved generated maps and all N training maps contribute. No high-N
cutoff and no real-map cap. PDFs use shared real/generated bins for each run, expanded to include generated
values outside the configured scaled-tanh bounds. All pixels are included; nothing is clipped. The generated
out-of-range fraction is printed. Real reference bounds are [-beta*sigma, alpha*sigma] for positive alpha, beta, sigma, as implemented by the loader. Power ratio = mean generated P(k)
divided by mean training P(k), not the mean of individual ratios.
Nearest matches shown later use centered pixel cosine over **all N maps**,
without D4 alignment; this is separate from PCA95.

```python
diagnostics = []
if runs:
    if not os.environ.get('SLURM_JOB_ID'):
        raise RuntimeError('Use an existing compute allocation for full-array diagnostics; no job is submitted here.')
    from simdiff_eval.io import iter_real_reference_batches_from_config, as_nchw
    from simdiff_eval.metrics import batch_power_spectra
    edges = np.linspace(-1,1,101)
    def unit(a):
        a = a.reshape(len(a),-1).astype(np.float64)
        a -= a.mean(axis=1,keepdims=True)
        norm = np.linalg.norm(a,axis=1,keepdims=True)
        assert (norm>0).all()
        return a/norm
    def check(a):
        assert a.shape[1:]==(1,128,128) and np.isfinite(a).all(), a.shape
        # Generated diffusion outputs need not lie within the tanh training range.
    import yaml
    for r in runs:
        data_cfg = yaml.safe_load(Path(r['config']).read_text())['data']
        assert data_cfg.get('normalization') == 'tanh', ('Unexpected normalization', r['config'])
        kw = data_cfg.get('norm_kwargs') or {}
        alpha, beta, sigma = [float(kw.get(key,1.0)) for key in ('alpha','beta','sigma')]
        assert np.isfinite([alpha,beta,sigma]).all() and min(alpha,beta,sigma)>0
        real_lo, real_hi = -beta*sigma, alpha*sigma
        tolerance = 1e-5*max(1.0,abs(real_lo),abs(real_hi))
        print(r['run_name'], 'normalization kwargs:', kw, 'real bounds:', (real_lo,real_hi))
        with np.load(r['sample'],allow_pickle=False) as f: generated=as_nchw(f['samples']).astype(np.float32)
        check(generated); assert len(generated)==512, (r['run_name'],len(generated))
        gp,k = batch_power_spectra(generated,nbins=40)
        edges = np.linspace(min(real_lo-tolerance,float(generated.min())), max(real_hi+tolerance,float(generated.max())),101)
        outside = float(np.mean((generated < real_lo) | (generated > real_hi)))
        print('Generated pixel range:', float(generated.min()), float(generated.max()), 'fraction outside configured real bounds:', outside)
        gh = np.histogram(generated,bins=edges)[0]
        assert gh.sum() == generated.size, 'Histogram lost generated pixels'
        rh=np.zeros(len(edges)-1,dtype=np.int64); ps=np.zeros(len(k)); count=0
        queries=unit(generated[:2]); best=np.full(2,-np.inf); matches=np.zeros_like(generated[:2]); ids=np.zeros(2,int)
        for batch in iter_real_reference_batches_from_config(r['config']):
            check(batch)
            assert batch.min()>=real_lo-tolerance and batch.max()<=real_hi+tolerance, ('Real reference outside configured tanh bounds',r['config'],float(batch.min()),float(batch.max()),real_lo,real_hi)
            rh += np.histogram(batch,bins=edges)[0]
            bp,bk = batch_power_spectra(batch,nbins=40)
            assert np.allclose(k,bk); ps += bp.sum(axis=0)
            cos=queries@unit(batch).T; idx=cos.argmax(axis=1); vals=cos[np.arange(2),idx]
            better=vals>best
            matches[better]=batch[idx[better]]; ids[better]=count+idx[better]; best[better]=vals[better]
            count+=len(batch)
        assert rh.sum() == count*128*128, 'Histogram lost real pixels'
        assert count==r['N'], (r['run_name'],count,r['N'])
        real_power=ps/count; gen_power=gp.mean(axis=0)
        diagnostics.append(dict(**r,k=k,edges=edges,outside_fraction=outside,real_pdf=rh/(rh.sum()*np.diff(edges)),
            gen_pdf=gh/(gh.sum()*np.diff(edges)),real_power=real_power,gen_power=gen_power,
            ratio=np.divide(gen_power,real_power,out=np.full_like(gen_power,np.nan),where=real_power>0),
            generated=generated[:2],matches=matches,match_ids=ids,cosines=best))
        print('Computed L%d N=%d: %d real / %d generated'%(r['depth'],r['N'],count,len(generated)))
else:
    print('PENDING: latest arrays unavailable here. No historical panels substituted.')
```

### 4. Full range: PDFs, P(k), and power ratios — N=64 through 32,768
Exactly two five-column pages per depth, including explicit missing-data slots.
The second page is **HIGH N: 2,048–32,768**. A computation that stopped early
cannot masquerade as a complete plot. PDF bins and vertical scales may differ
between runs; compare generated and reference within each panel. These are
model-space statistics of all 512 generated maps and all N training maps,
not physical-unit HI power or an independent held-out reference.

```python
# ALL-N-32768-v1: explicit slots, never a five-size truncation.
expected_keys = {(r['depth'], r['N']) for r in runs}
computed_keys = {(r['depth'], r['N']) for r in diagnostics}
assert computed_keys == expected_keys, 'Computation incomplete: rerun the full-reference cell before plotting.'
assert len(computed_keys) == len(diagnostics), 'Duplicate diagnostics'
by_key = {(r['depth'], r['N']): r for r in diagnostics}
for depth in (8,12,16):
    for sizes, label in ((EXPECTED_SIZES[:5], 'LOW N'), (EXPECTED_SIZES[5:], 'HIGH N')):
        fig, axs = plt.subplots(3, 5, figsize=(17, 8.5), squeeze=False, constrained_layout=True)
        for j, size in enumerate(sizes):
            axs[0,j].set_title('N = {:,} ($2^{{{}}}$)'.format(size, int(np.log2(size))))
            r = by_key.get((depth, size))
            if r is None:
                reason = next((x['status'] for x in inventory if x['depth']==depth and x['N']==size),
                              'No verified entry in selected plans')
                for ax in axs[:,j]:
                    ax.text(.5, .5, 'NOT AVAILABLE\n' + reason, ha='center', va='center',
                            transform=ax.transAxes, fontsize=9, wrap=True)
                    ax.set_xticks([]); ax.set_yticks([])
                continue
            x = (r['edges'][1:] + r['edges'][:-1])/2
            axs[0,j].semilogy(x, r['real_pdf'], color='black', label='Exact training')
            axs[0,j].semilogy(x, r['gen_pdf'], color='#226699', ls='--', label='Generated')
            axs[0,j].set_xlabel('Normalized pixel value')
            positive = r['k'] > 0
            axs[1,j].loglog(r['k'][positive], r['real_power'][positive], color='black')
            axs[1,j].loglog(r['k'][positive], r['gen_power'][positive], color='#226699', ls='--')
            axs[2,j].plot(r['k'], r['ratio'], color='#226699')
            axs[2,j].axhline(1, color='.4', ls='--')
            for i in (1,2): axs[i,j].set_xlabel('Radial k (grid units)')
            for ax in axs[:,j]: ax.grid(alpha=.15)
            axs[0,j].legend(fontsize=8)
        axs[0,0].set_ylabel('Pixel probability density')
        axs[1,0].set_ylabel('Mean model-space P(k)')
        axs[2,0].set_ylabel('Generated / training P(k)')
        fig.suptitle('DiT-L{} | zero-init 300k | {}: N={:,}–{:,}'.format(depth,label,sizes[0],sizes[-1]))
        plt.show()
        plt.close(fig)
        for size in sizes:
            r = by_key.get((depth,size))
            if r: print('L%d N=%d | %s | %s' % (depth,size,r['sample'],r['config']))
print('Displayed all 30 depth/size slots through 2**15 = 32,768; absent data are labelled, not filled.')
```

### 5. Images and closest matches at every available N
Fixed generated indices 0 and 1; closest matches search all N training maps.
Generated and matched images share an unclipped range within each run;
colorbars show the scale. Scales can differ across runs. These are centered
pixel-cosine matches without D4 alignment, not the PCA95 score.

```python
assert {(r['depth'],r['N']) for r in diagnostics} == {(r['depth'],r['N']) for r in runs}, 'Incomplete computation'
for r in diagnostics:
    fig,axs=plt.subplots(2,3,figsize=(9,5.5),constrained_layout=True)
    lo = min(float(r['generated'].min()), float(r['matches'].min()))
    hi = max(float(r['generated'].max()), float(r['matches'].max()))
    diff_hi = max(float(abs(r['generated']-r['matches']).max()),1e-12)
    for i in range(2):
        a,b=r['generated'][i,0],r['matches'][i,0]
        for j,(img,title) in enumerate(zip([a,b,abs(a-b)],['Generated','Nearest training','Absolute difference'])):
            artist=axs[i,j].imshow(img,cmap='magma' if j==2 else 'viridis',
                                  vmin=0 if j==2 else lo,vmax=diff_hi if j==2 else hi)
            axs[i,j].set_title(title); axs[i,j].set_xticks([]); axs[i,j].set_yticks([])
            if i==0: fig.colorbar(artist,ax=axs[:,j],shrink=.7)
        axs[i,1].set_xlabel('row %d; cosine %.4f'%(r['match_ids'][i],r['cosines'][i]))
    fig.suptitle('L%d | N=%s | verified 300k'%(r['depth'],format(r['N'],',')))
    plt.show(); plt.close(fig)
    print(r['run_name'], '|', r['sample'])
```
